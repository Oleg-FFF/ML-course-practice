1. Задана функція залежності оцінки від кількості годин, витрачених на навчання: $ f(x) = 10x - 5 $.
  1. З допомогою Python обчисліть похідну цієї функції.
  2. Використовуйте бібліотеку `matplotlib` для побудови графіків функції та її похідної в діапазоні значень від 0 до 10 годин. Проаналізуйте, як зміна кількості годин вивчення впливає на оцінку.



```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 100)
f = 10 * x - 5
df = 10 * np.ones_like(x)

plt.figure(figsize=(10, 5))
plt.plot(x, f, label='f(x) = 10 * x - 5', color='blue')
plt.plot(x, df, label="f'(x) = 10 * np.ones_like(x)", color='green')

plt.title("Графік функції f(x) = 10x - 5 та її похідної")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)

plt.show()

# Висновок: Для даної моделі - кожна година навчання дає 10 балів. Це в теорії, на практиці є максимальна кількість балів, втомлюваність і т.д.
```

2. Використовуючи Python і `SymPy`, обчисліть похідні для функцій $ f(x) = \sin(x) $ та $ f(x) = \ln(x) $. Зобразіть ці функції та їхні похідні за допомогою `matplotlib`.



```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

x = sp.symbols('x')

f_sin = sp.sin(x)
df_sin = sp.diff(f_sin, x)

print("f(x) = sin(x), f'(x) =", df_sin)

f_ln = sp.log(x)
df_ln = sp.diff(f_ln, x)
print("f(x) = ln(x), f'(x) =", df_ln)

x_sin = np.linspace(-2 * np.pi, 2 * np.pi, 400)
y_sin = np.sin(x_sin)
y_cos = np.cos(x_sin)

x_ln = np.linspace(0.1, 5, 400)
y_ln = np.log(x_ln)
y_ln_df = 1 / x_ln

plt.figure()
plt.plot(x_sin, y_sin, label='sin(x)')
plt.plot(x_sin, y_cos, label="cos(x) = d/dx sin(x)")
plt.xlabel('x')
plt.ylabel('y')
plt.title('f(x) = sin(x) та її похідна')
plt.legend()
plt.grid(True)
plt.show()

plt.figure()
plt.plot(x_ln, y_ln, label='ln(x)')
plt.plot(x_ln, y_ln_df, label="1/x = d/dx ln(x)")
plt.xlabel('x')
plt.ylabel('y')
plt.title('f(x) = ln(x) та її похідна')
plt.legend()
plt.grid(True)
plt.show()
```


3. Загальні витрати компанії від обсягу виробництва задані функцією $ f(x) = x^2 - 5x + 1 $, де $ x $ — кількість вироблених одиниць. Обчисліть похідну цієї функції з допомогою SymPy та візуалізуйте значення функції і її похідної на відрізку від 0 до 100.

```python
x = sp.symbols('x')
f = x ** 2 - 5 * x + 1
df_f = sp.diff(f, x)
print("f(x) = x^2 - 5 * x + 1, f'(x)=", df_f)

x_line = np.linspace(0, 100, 500)
x_f = x_line ** 2 - 5 * x_line + 1
x_df = 2 * x_line - 5

plt.figure()
plt.plot(x_line, x_f, label='x^2 - 5 * x + 1')
plt.plot(x_line, x_df, label="2*x - 5")
plt.xlabel('x')
plt.ylabel('y')
plt.title('f(x) = x^2 - 5 * x + 1 та її похідна')
plt.legend()
plt.grid(True)
plt.show()
```
