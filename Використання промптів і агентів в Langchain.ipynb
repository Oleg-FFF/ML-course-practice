Тепер ми хочемо оновити попередній фукнціонал так, аби в промпт ми могли передавати тему як параметр. Для цього скористайтесь `PromptTemplate` з `langchain` і реалізуйте параметризований промпт та виклик моделі з ним.

Запустіть оновлений функціонал (промпт + модел) для пояснень про теми
- "Баєсівські методи в машинному навчанні"

Виведіть результати відпрацювання моделі на екран.

````python
from pydantic import BaseModel, Field
from typing import List

from langchain_openai import ChatOpenAI
from langchain.agents import create_agent
from langchain_community.tools import DuckDuckGoSearchRun


class Paper(BaseModel):
    title: str
    authors: str
    summary: str


class PaperList(BaseModel):
    papers: List[Paper]


llm = ChatOpenAI(model="gpt-4.1-mini", temperature=0)
search_tool = DuckDuckGoSearchRun()

agent = create_agent(
    model=llm,
    tools=[search_tool],
    system_prompt=(
        "Ти науковий асистент. "
        "Шукай лише сторінки з науковими публікаціями або репозиторіями статей. "
        "Не вигадуй авторів. Якщо автора немає в джерелі, пиши 'Невідомо'."
    ),
    response_format=PaperList,
)

result = agent.invoke(
    {
        "messages": [
            {
                "role": "user",
                "content": "Знайди 5 публікацій про artificial intelligence"
            }
        ]
    }
)

print(result["structured_response"])
````

````python
import os
import requests
import time

from pydantic import BaseModel, Field
from typing import List

from langchain_openai import ChatOpenAI
from langchain.agents import create_agent
from langchain.tools import tool


os.environ["OPENAI_API_KEY"] = api_key


class Paper(BaseModel):
    title: str = Field(description="Назва публікації")
    authors: str = Field(description="Автори через кому")
    summary: str = Field(description="Короткий опис публікації")


class PaperList(BaseModel):
    papers: List[Paper]


def safe_request(url, params):
    for i in range(20):
        resp = requests.get(url, params=params)
        
        if resp.status_code == 429:
            print("Rate limit... sleep")
            time.sleep(2 * (i + 1))
            continue
        
        resp.raise_for_status()
        return resp
    
    raise Exception("Too many retries")

@tool
def search_semantic_scholar(topic: str) -> str:
    """Шукає наукові публікації за темою і повертає сирі дані."""
    url = "https://api.semanticscholar.org/graph/v1/paper/search"
    params = {
        "query": topic,
        "limit": 5,
        "fields": "title,authors,abstract,year,url"
    }

    resp = safe_request(url, params)
    resp.raise_for_status()
    data = resp.json()

    rows = []
    for paper in data.get("data", []):
        authors = ", ".join(a["name"] for a in paper.get("authors", []))
        rows.append({
            "title": paper.get("title", ""),
            "authors": authors,
            "summary": (paper.get("abstract") or "")[:500]
        })

    return str(rows)


llm = ChatOpenAI(
    model="gpt-4.1-mini",
    temperature=0,
)

agent = create_agent(
    model=llm,
    tools=[search_semantic_scholar],
    system_prompt=(
        "Ти науковий асистент. "
        "Для пошуку статей обов'язково використовуй tool search_semantic_scholar. "
        "Поверни рівно 5 останніх або найбільш релевантних публікацій."
    ),
    response_format=PaperList,
)

result = agent.invoke(
    {
        "messages": [
            {
                "role": "user",
                "content": "Знайди 5 наукових публікацій на тему: Штучний інтелект"
            }
        ]
    }
)

print("STRUCTURED RESPONSE:")
print(result["structured_response"])

print("\nMESSAGES:")
for msg in result["messages"]:
    print(type(msg).__name__, "->", getattr(msg, "content", ""))
    print("-----")
````

2. Створіть запит до агента, що містить чітке завдання – видати результат бізнес аналізу або написати, що він не може цього зробити і запит користувача (просто може бути все одним повідомлленням).

````python
from langchain_openai import ChatOpenAI
from langchain.agents import create_agent
from langchain_experimental.tools import PythonREPLTool

python_tool = PythonREPLTool()

llm = ChatOpenAI(
    model="gpt-4o-mini",
    temperature=0
)

search_tool = DuckDuckGoSearchRun()

agent = create_agent(
    model=llm,
    tools=[
        python_tool,
        search_tool
    ],
    system_prompt="""
    Ти помічник, який допомагає вирішувати задачі бізнес-аналітики.

    - Для пошуку інформації використовуй веб пошук
    - Для обробки даних або обчислень використовуй python_repl
    """
)

user_prompt = "Ми експортуємо апельсини з Бразилії. В 2022 експортували 200т, в 2023 - 190т, в 2024 - 210т, в 2025 - 220т. Зроби оцінку скільки ми зможемо експортувати апельсинів в 2026 враховуючи погодні умови в Бразилії і попит на апельсини в світі виходячи з економічної ситуації."

result = agent.invoke(
    {
        "messages": [
            {
                "role": "user",
                "content": user_prompt
            }
        ]
    }
)

for msg in result["messages"]:
    print(f"\n=== {type(msg).__name__} ===")
    print(getattr(msg, "content", ""))
````
